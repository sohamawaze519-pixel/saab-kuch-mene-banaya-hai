```python
import re
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from bs4 import BeautifulSoup
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, f1_score

warnings.filterwarnings('ignore')
print("Libraries imported successfully!")
```

```python
# Download required NLTK resources
for resource in ['stopwords', 'punkt', 'wordnet', 'omw-1.4']:
    try:
        nltk.data.find(f'corpora/{resource}' if resource in ['stopwords', 'wordnet', 'omw-1.4'] else f'tokenizers/{resource}')
    except LookupError:
        nltk.download(resource, quiet=True)

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def clean_text(text):
    """Clean and preprocess individual text attributes."""
    if not isinstance(text, str) or not text.strip():
        return ""
    
    # 1. Strip HTML tags
    text = BeautifulSoup(text, "html.parser").get_text(separator=" ")
    
    # 2. Lowercase and remove non-alphabetic characters
    text = re.sub(r'[^a-zA-Z\s]', ' ', text.lower())
    
    # 3. Tokenize
    tokens = word_tokenize(text)
    
    # 4. Remove stopwords and short noisy tokens, apply lemmatization
    cleaned = [
        lemmatizer.lemmatize(word) 
        for word in tokens 
        if word not in stop_words and len(word) > 2
    ]
    
    return " ".join(cleaned)

print("Text cleaning function ready.")
```
